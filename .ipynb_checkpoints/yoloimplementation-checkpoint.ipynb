```python
import cv2
from ultralytics import YOLO

def main():
    # 1. Load the lightweight YOLO model for high real-time performance
    model = YOLO("yolo11n.pt") 

    # 2. Initialize video source
    # Change "sample_video.mp4" to 0 to use your system's default live webcam feed
    video_source = "sample_video.mp4" 
    cap = cv2.VideoCapture(0)

    if not cap.isOpened():
        print("Error: Could not open video source.")
        return

    print("Processing video. Press 'q' to quit.")

    while cap.isOpened():
        success, frame = cap.read()
        
        if not success:
            print("Video stream finished or failed to read frame.")
            break

        # 3. Run YOLO tracking on the current frame
        # 'persist=True' keeps track of object IDs across consecutive frames
        results = model.track(frame, persist=True)

        # 4. Extract annotated frame (with default bounding boxes and labels)
        annotated_frame = results[0].plot()

        # 5. Display the output frame in a window
        cv2.imshow("YOLO Real-Time Detection", annotated_frame)

        # Break the loop safely if the user presses the 'q' key
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    # 6. Clean up resources and close windows
    cap.release()
    cv2.destroyAllWindows()

if __name__ == "__main__":
    main()
```
